```python
import sys
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
import torchvision.transforms as transforms
from torchvision.models import vgg16
import pandas as pd
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm
import nltk
from pathlib import Path

try:
    nltk.data.find("tokenizers/punkt")
except LookupError:
    nltk.download("punkt")

if "project_root" not in globals():
    notebook_dir = Path.cwd()
    project_root = notebook_dir.parent
    print(f"project_root was not defined — set to: {project_root}")

# Inline implementations instead of importing from .py files


class Tokenizer:
    def __init__(self, vocab_size=5000):
        self.vocab_size = vocab_size
        self.word2idx = {}
        self.idx2word = {}
        self.special_tokens = ["<pad>", "<start>", "<end>", "<unk>"]
        self._init_special_tokens()

    def _init_special_tokens(self):
        for token in self.special_tokens:
            idx = len(self.word2idx)
            self.word2idx[token] = idx
            self.idx2word[idx] = token

    def build_vocab(self, sentences):
        # Clean and tokenize sentences (handle Persian and remove emojis)
        import re

        cleaned_sentences = []
        for sent in sentences:
            # Remove emojis
            sent = re.sub(r"[^\w\s\u0600-\u06FF]", "", sent)  # Keep Persian characters
            cleaned_sentences.append(sent)

        # Build vocabulary
        from collections import Counter

        all_words = []
        for sent in cleaned_sentences:
            words = sent.split()
            all_words.extend(words)

        word_counts = Counter(all_words)
        most_common = word_counts.most_common(
            self.vocab_size - len(self.special_tokens)
        )

        for word, _ in most_common:
            if word not in self.word2idx:
                idx = len(self.word2idx)
                self.word2idx[word] = idx
                self.idx2word[idx] = word

    def encode(self, sentence):
        # Clean sentence
        import re

        sentence = re.sub(r"[^\w\s\u0600-\u06FF]", "", sentence)
        words = sentence.split()
        tokens = [self.word2idx.get("<start>", 1)]
        for word in words:
            tokens.append(self.word2idx.get(word, self.word2idx.get("<unk>", 3)))
        tokens.append(self.word2idx.get("<end>", 2))
        return tokens

    def decode(self, tokens):
        words = []
        for token in tokens:
            if token == self.word2idx.get("<end>", 2):
                break
            if token not in [
                self.word2idx.get("<start>", 1),
                self.word2idx.get("<pad>", 0),
            ]:
                words.append(self.idx2word.get(token, "<unk>"))
        return " ".join(words)

    def __len__(self):
        return len(self.word2idx)


# EncoderCNN with VGG16 instead of ResNet
import torch.nn as nn
from torchvision.models import vgg16


class EncoderCNN(nn.Module):
    def __init__(self, encoded_dim=512, freeze_backbone=True):
        super(EncoderCNN, self).__init__()
        # Use VGG16 as per assignment requirements
        vgg = vgg16(pretrained=True)
        # Remove the last classification layer
        self.features = nn.Sequential(*list(vgg.features.children()))
        self.adaptive_pool = nn.AdaptiveAvgPool2d((7, 7))
        self.linear = nn.Linear(512 * 7 * 7, encoded_dim)
        self.bn = nn.BatchNorm1d(encoded_dim, momentum=0.01)
        self.dropout = nn.Dropout(0.5)

        if freeze_backbone:
            for param in self.features.parameters():
                param.requires_grad = False

    def forward(self, images):
        features = self.features(images)
        features = self.adaptive_pool(features)
        features = features.view(features.size(0), -1)
        features = self.linear(features)
        features = self.bn(features)
        features = self.dropout(features)
        return features


class DecoderLSTM(nn.Module):
    def __init__(self, vocab_size, embed_dim, decoder_dim, encoder_dim):
        super(DecoderLSTM, self).__init__()
        self.vocab_size = vocab_size
        self.embed_dim = embed_dim
        self.decoder_dim = decoder_dim
        self.encoder_dim = encoder_dim

        self.embedding = nn.Embedding(vocab_size, embed_dim)
        self.lstm = nn.LSTMCell(embed_dim + encoder_dim, decoder_dim, bias=True)
        self.linear = nn.Linear(decoder_dim, vocab_size)
        self.dropout = nn.Dropout(0.5)

        # Attention
        self.attention_linear = nn.Linear(decoder_dim, encoder_dim)
        self.attention_softmax = nn.Softmax(dim=1)

    def forward(self, encoder_out, captions):
        batch_size = captions.size(0)
        seq_len = captions.size(1)

        # Initialize hidden states
        h = torch.zeros(batch_size, self.decoder_dim).to(captions.device)
        c = torch.zeros(batch_size, self.decoder_dim).to(captions.device)

        outputs = []
        attentions = []

        for t in range(seq_len):
            if t == 0:
                # First input is <start> token
                word_embed = self.embedding(captions[:, t])
            else:
                # Teacher forcing: use ground truth
                word_embed = self.embedding(captions[:, t])

            # Attention
            att = self.attention_linear(h)
            att = torch.bmm(att.unsqueeze(1), encoder_out.permute(0, 2, 1)).squeeze(1)
            att_weights = self.attention_softmax(att)
            context = torch.bmm(att_weights.unsqueeze(1), encoder_out).squeeze(1)

            # LSTM input
            lstm_input = torch.cat([word_embed, context], dim=1)
            h, c = self.lstm(lstm_input, (h, c))

            # Output
            output = self.linear(self.dropout(h))
            outputs.append(output)
            attentions.append(att_weights)

        outputs = torch.stack(outputs, dim=1)
        attentions = torch.stack(attentions, dim=1)
        return outputs, attentions

    def greedy_decode(self, encoder_out, max_len=40):
        batch_size = encoder_out.size(0)

        # Initialize
        h = torch.zeros(batch_size, self.decoder_dim).to(encoder_out.device)
        c = torch.zeros(batch_size, self.decoder_dim).to(encoder_out.device)

        # Start with <start> token
        start_token = 1  # Assuming <start> is 1
        current_word = torch.full((batch_size,), start_token, dtype=torch.long).to(
            encoder_out.device
        )

        outputs = []
        attentions = []

        for t in range(max_len):
            word_embed = self.embedding(current_word)

            # Attention
            att = self.attention_linear(h)
            att = torch.bmm(att.unsqueeze(1), encoder_out.permute(0, 2, 1)).squeeze(1)
            att_weights = self.attention_softmax(att)
            context = torch.bmm(att_weights.unsqueeze(1), encoder_out).squeeze(1)

            # LSTM
            lstm_input = torch.cat([word_embed, context], dim=1)
            h, c = self.lstm(lstm_input, (h, c))

            # Output
            output = self.linear(self.dropout(h))
            outputs.append(output)
            attentions.append(att_weights)

            # Get next word
            current_word = output.argmax(dim=1)

            # Stop if all sequences ended
            if (current_word == 2).all():  # Assuming <end> is 2
                break

        outputs = torch.stack(outputs, dim=1)
        attentions = torch.stack(attentions, dim=1)
        return outputs, attentions


# Dataset class
class ImageCaptionDataset(torch.utils.data.Dataset):
    def __init__(self, df, images_root, tokenizer, transforms=None):
        self.df = df
        self.images_root = Path(images_root)
        self.tokenizer = tokenizer
        self.transforms = transforms

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        img_path = self.images_root / row["image"]

        # Load image
        image = Image.open(img_path).convert("RGB")
        if self.transforms:
            image = self.transforms(image)

        # Tokenize caption
        caption = self.tokenizer.encode(row["caption"])

        return image, torch.tensor(caption), len(caption)


# Collate function
def collate_fn(batch, tokenizer):
    images, captions, lengths = zip(*batch)

    # Pad captions
    max_len = max(lengths)
    padded_captions = []
    for cap in captions:
        if len(cap) < max_len:
            pad_len = max_len - len(cap)
            cap = torch.cat(
                [cap, torch.full((pad_len,), tokenizer.word2idx.get("<pad>", 0))]
            )
        padded_captions.append(cap)

    return torch.stack(images), torch.stack(padded_captions), torch.tensor(lengths)


# BLEU score calculation
from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction


def calculate_bleu_score(generated, reference):
    # Calculate BLEU-1 to BLEU-4
    reference_tokens = reference.split()
    generated_tokens = generated.split()

    smoothing = SmoothingFunction().method1

    bleu1 = sentence_bleu(
        [reference_tokens],
        generated_tokens,
        weights=(1, 0, 0, 0),
        smoothing_function=smoothing,
    )
    bleu2 = sentence_bleu(
        [reference_tokens],
        generated_tokens,
        weights=(0.5, 0.5, 0, 0),
        smoothing_function=smoothing,
    )
    bleu3 = sentence_bleu(
        [reference_tokens],
        generated_tokens,
        weights=(0.33, 0.33, 0.33, 0),
        smoothing_function=smoothing,
    )
    bleu4 = sentence_bleu(
        [reference_tokens],
        generated_tokens,
        weights=(0.25, 0.25, 0.25, 0.25),
        smoothing_function=smoothing,
    )

    return bleu1, bleu2, bleu3, bleu4


def calculate_caption_length(caption):
    return len(caption.split())


# Seed everything function
def seed_everything(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False


plt.style.use("default")
sns.set_palette("husl")
print("All imports successful!")
```

```python
# Q1: Corrected Tokenizer, Encoder (VGG), Attention, Decoder, Dataset and BLEU utilities
import re

try:
    import emoji

    _has_emoji = True
except Exception:
    _has_emoji = False


class AdvancedTokenizer:
    def __init__(self, vocab_size=5000):
        self.vocab_size = vocab_size
        self.word2idx = {"<pad>": 0, "<start>": 1, "<end>": 2, "<unk>": 3}
        self.idx2word = {v: k for k, v in self.word2idx.items()}
        self.word_counts = {}

    def preprocess_text(self, text):
        text = str(text)
        if _has_emoji:
            text = emoji.replace_emoji(text, replace="")
        else:
            text = re.sub(r"[^\w\s\u0600-\u06FF]", "", text)
        text = text.lower().strip()
        return text

    def build_vocab(self, captions):
        for caption in captions:
            clean_caption = self.preprocess_text(caption)
            tokens = clean_caption.split()
            for token in tokens:
                self.word_counts[token] = self.word_counts.get(token, 0) + 1

        most_common = sorted(self.word_counts.items(), key=lambda x: x[1], reverse=True)
        for idx, (word, _) in enumerate(most_common[: self.vocab_size - 4]):
            self.word2idx[word] = idx + 4
            self.idx2word[idx + 4] = word

    def __call__(self, text):
        clean_text = self.preprocess_text(str(text))
        tokens = clean_text.split()
        return [self.word2idx.get(token, self.word2idx["<unk>"]) for token in tokens]

    def encode(self, sentence):
        tokens = (
            [self.word2idx["<start>"]]
            + self.__call__(sentence)
            + [self.word2idx["<end>"]]
        )
        return tokens

    def decode(self, token_list):
        words = []
        for t in token_list:
            if t == self.word2idx["<end>"]:
                break
            if t not in (self.word2idx["<pad>"], self.word2idx["<start>"]):
                words.append(self.idx2word.get(t, "<unk>"))
        return " ".join(words)

    def __len__(self):
        return len(self.word2idx)


# Encoder, Attention, Decoder (VGG + Bahdanau-style)
import torch.nn as nn
from torchvision.models import vgg16, VGG16_Weights


class EncoderVGG(nn.Module):
    def __init__(self, encoded_dim=512, freeze_backbone=True):
        super(EncoderVGG, self).__init__()
        vgg = vgg16(weights=VGG16_Weights.IMAGENET1K_V1)
        self.features = vgg.features
        if freeze_backbone:
            for p in self.features.parameters():
                p.requires_grad = False

    def forward(self, images):
        feats = self.features(images)  # B x 512 x 7 x 7
        b, c, h, w = feats.size()
        feats = feats.permute(0, 2, 3, 1).view(b, h * w, c)  # B x 49 x 512
        return feats


class Attention(nn.Module):
    def __init__(self, encoder_dim, decoder_dim, attention_dim):
        super(Attention, self).__init__()
        self.encoder_att = nn.Linear(encoder_dim, attention_dim)
        self.decoder_att = nn.Linear(decoder_dim, attention_dim)
        self.full_att = nn.Linear(attention_dim, 1)
        self.relu = nn.ReLU()
        self.softmax = nn.Softmax(dim=1)

    def forward(self, encoder_out, decoder_hidden):
        att1 = self.encoder_att(encoder_out)  # B x L x attn
        att2 = self.decoder_att(decoder_hidden).unsqueeze(1)  # B x 1 x attn
        att = self.full_att(self.relu(att1 + att2)).squeeze(2)  # B x L
        alpha = self.softmax(att)
        context = (encoder_out * alpha.unsqueeze(2)).sum(dim=1)
        return context, alpha


class DecoderLSTM_V2(nn.Module):
    def __init__(self, vocab_size, embed_dim, decoder_dim, encoder_dim, attention_dim):
        super(DecoderLSTM_V2, self).__init__()
        self.embedding = nn.Embedding(vocab_size, embed_dim)
        self.lstm = nn.LSTMCell(embed_dim + encoder_dim, decoder_dim)
        self.attention = Attention(encoder_dim, decoder_dim, attention_dim)
        self.fc = nn.Linear(decoder_dim, vocab_size)
        self.dropout = nn.Dropout(0.5)
        self.init_h = nn.Linear(encoder_dim, decoder_dim)
        self.init_c = nn.Linear(encoder_dim, decoder_dim)

    def init_hidden_state(self, encoder_out):
        mean_enc = encoder_out.mean(dim=1)
        h = self.init_h(mean_enc)
        c = self.init_c(mean_enc)
        return h, c

    def forward(self, encoder_out, captions, teacher_forcing_ratio=0.5):
        batch_size = encoder_out.size(0)
        seq_len = captions.size(1)
        vocab_size = self.fc.out_features

        outputs = torch.zeros(batch_size, seq_len, vocab_size).to(encoder_out.device)
        alphas = torch.zeros(batch_size, seq_len, encoder_out.size(1)).to(
            encoder_out.device
        )

        embeddings = self.embedding(captions)
        h, c = self.init_hidden_state(encoder_out)
        x = embeddings[:, 0, :]

        for t in range(1, seq_len):
            context, alpha = self.attention(encoder_out, h)
            alphas[:, t, :] = alpha
            lstm_input = torch.cat([x, context], dim=1)
            h, c = self.lstm(lstm_input, (h, c))
            out = self.fc(self.dropout(h))
            outputs[:, t, :] = out

            is_teacher = np.random.random() < teacher_forcing_ratio
            top1 = out.argmax(1)
            x = embeddings[:, t, :] if is_teacher else self.embedding(top1)

        return outputs, alphas

    def greedy_decode(self, encoder_out, max_len=40):
        batch_size = encoder_out.size(0)
        h, c = self.init_hidden_state(encoder_out)
        current = torch.full((batch_size,), 1, dtype=torch.long).to(encoder_out.device)
        outputs = []
        alphas = []
        for t in range(max_len):
            emb = self.embedding(current)
            context, alpha = self.attention(encoder_out, h)
            h, c = self.lstm(torch.cat([emb, context], dim=1), (h, c))
            out = self.fc(self.dropout(h))
            outputs.append(out)
            alphas.append(alpha)
            current = out.argmax(dim=1)
            if (current == 2).all():
                break
        outputs = torch.stack(outputs, dim=1)
        alphas = torch.stack(alphas, dim=1)
        return outputs, alphas


# Dataset and collate
class ImageCaptionDataset(torch.utils.data.Dataset):
    def __init__(self, df, images_root, tokenizer, transforms=None):
        self.df = df
        self.images_root = Path(images_root)
        self.tokenizer = tokenizer
        self.transforms = transforms

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        img_path = self.images_root / row["image"]
        image = Image.open(img_path).convert("RGB")
        if self.transforms:
            image = self.transforms(image)
        caption_tokens = torch.tensor(self.tokenizer.encode(row["caption"]))
        return image, caption_tokens, len(caption_tokens)


def collate_fn(batch, tokenizer):
    images, captions, lengths = zip(*batch)
    max_len = max(lengths)
    padded = []
    for cap in captions:
        if len(cap) < max_len:
            pad_len = max_len - len(cap)
            cap = torch.cat(
                [
                    cap,
                    torch.full(
                        (pad_len,), tokenizer.word2idx.get("<pad>", 0), dtype=torch.long
                    ),
                ]
            )
        padded.append(cap)
    return torch.stack(images), torch.stack(padded), torch.tensor(lengths)


# BLEU utilities
from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction


def calculate_bleu_metrics(generated_captions, reference_captions):
    bleu1, bleu2, bleu3, bleu4 = [], [], [], []
    smooth_fn = SmoothingFunction().method1
    for gen, ref in zip(generated_captions, reference_captions):
        ref_tokens = [re.sub(r"[^\w\s\u0600-\u06FF]", "", ref).split()]
        gen_tokens = re.sub(r"[^\w\s\u0600-\u06FF]", "", gen).split()
        bleu1.append(
            sentence_bleu(
                ref_tokens,
                gen_tokens,
                weights=(1, 0, 0, 0),
                smoothing_function=smooth_fn,
            )
        )
        bleu2.append(
            sentence_bleu(
                ref_tokens,
                gen_tokens,
                weights=(0.5, 0.5, 0, 0),
                smoothing_function=smooth_fn,
            )
        )
        bleu3.append(
            sentence_bleu(
                ref_tokens,
                gen_tokens,
                weights=(0.33, 0.33, 0.33, 0),
                smoothing_function=smooth_fn,
            )
        )
        bleu4.append(
            sentence_bleu(
                ref_tokens,
                gen_tokens,
                weights=(0.25, 0.25, 0.25, 0.25),
                smoothing_function=smooth_fn,
            )
        )
    return {
        "BLEU-1": np.mean(bleu1),
        "BLEU-2": np.mean(bleu2),
        "BLEU-3": np.mean(bleu3),
        "BLEU-4": np.mean(bleu4),
    }


print("Inserted corrected Q1 implementations (tokenizer, encoder, decoder, BLEU)")
```

## 6. Training Loop

Implement the complete training loop with loss computation, optimization, and validation.


```python
criterion = nn.CrossEntropyLoss(ignore_index=tokenizer.word2idx["<pad>"])

encoder_optimizer = optim.Adam(encoder.parameters(), lr=CONFIG["learning_rate"])
decoder_optimizer = optim.Adam(decoder.parameters(), lr=CONFIG["learning_rate"])

encoder_scheduler = optim.lr_scheduler.ReduceLROnPlateau(
    encoder_optimizer, mode="min", factor=0.5, patience=3, verbose=True
)
decoder_scheduler = optim.lr_scheduler.ReduceLROnPlateau(
    decoder_optimizer, mode="min", factor=0.5, patience=3, verbose=True
)
```

```python
print("Training components initialized:")
print(f"- Loss function: {criterion}")
print(f"- Encoder optimizer: {encoder_optimizer}")
print(f"- Decoder optimizer: {decoder_optimizer}")
print(f"- Learning rate schedulers: ReduceLROnPlateau")
```

```python
train_losses = []
val_losses = []
learning_rates = []

teacher_forcing_ratio = CONFIG["teacher_forcing_ratio"]

for epoch in range(CONFIG["num_epochs"]):
    print(f"\nEpoch {epoch+1}/{CONFIG['num_epochs']}")

    encoder.train()
    decoder.train()

    epoch_train_loss = 0
    train_batches = 0

    train_pbar = tqdm(train_loader, desc=f"Train Epoch {epoch+1}")
    for images, captions, lengths in train_pbar:
        images = images.to(device)
        captions = captions.to(device)

        encoder_optimizer.zero_grad()
        decoder_optimizer.zero_grad()

        features = encoder(images)
        outputs, attentions = decoder(captions, features)

        targets = captions[:, 1:]
        outputs = outputs[:, :-1, :]

        loss = criterion(
            outputs.contiguous().view(-1, len(tokenizer)), targets.contiguous().view(-1)
        )

        # Backward pass
        loss.backward()

        torch.nn.utils.clip_grad_norm_(encoder.parameters(), max_norm=1.0)
        torch.nn.utils.clip_grad_norm_(decoder.parameters(), max_norm=1.0)

        # Optimization step
        encoder_optimizer.step()
        decoder_optimizer.step()

        epoch_train_loss += loss.item()
        train_batches += 1

        train_pbar.set_postfix({"loss": f"{loss.item():.4f}"})

    avg_train_loss = epoch_train_loss / train_batches
    train_losses.append(avg_train_loss)

    # Validation phase
    encoder.eval()
    decoder.eval()

    epoch_val_loss = 0
    val_batches = 0

    with torch.no_grad():
        val_pbar = tqdm(val_loader, desc=f"Val Epoch {epoch+1}")
        for images, captions, lengths in val_pbar:
            images = images.to(device)
            captions = captions.to(device)

            features = encoder(images)
            outputs, attentions = decoder(captions, features)

            targets = captions[:, 1:]
            outputs = outputs[:, :-1, :]

            loss = criterion(
                outputs.contiguous().view(-1, len(tokenizer)),
                targets.contiguous().view(-1),
            )

            epoch_val_loss += loss.item()
            val_batches += 1

            val_pbar.set_postfix({"loss": f"{loss.item():.4f}"})

    avg_val_loss = epoch_val_loss / val_batches
    val_losses.append(avg_val_loss)

    # Learning rate scheduling
    encoder_scheduler.step(avg_val_loss)
    decoder_scheduler.step(avg_val_loss)

    # Record learning rate
    current_lr = encoder_optimizer.param_groups[0]["lr"]
    learning_rates.append(current_lr)

    # Update teacher forcing ratio (decay over time)
    teacher_forcing_ratio = max(0.0, teacher_forcing_ratio - 0.02)

    print(f"Epoch {epoch+1} Summary:")
    print(f"  Train Loss: {avg_train_loss:.4f}")
    print(f"  Val Loss: {avg_val_loss:.4f}")
    print(f"  Learning Rate: {current_lr:.6f}")
    print(f"  Teacher Forcing Ratio: {teacher_forcing_ratio:.3f}")

print("\nTraining completed!")
```

## 7. Inference and Caption Generation

Implement inference functions to generate captions for new images and visualize attention weights.


```python
# Inference functions
print("=== Inference Functions ===")


def generate_caption(
    encoder, decoder, image_path, tokenizer, max_length=50, device="cpu"
):
    """
    Generate a caption for a given image.

    Args:
        encoder: Trained encoder model
        decoder: Trained decoder model
        image_path: Path to the input image
        tokenizer: Tokenizer for text processing
        max_length: Maximum caption length
        device: Device to run inference on

    Returns:
        caption: Generated caption string
        attention_weights: Attention weights for visualization (not implemented in this version)
    """
    # Load and preprocess image
    transform = transforms.Compose(
        [
            transforms.Resize(CONFIG["image_size"]),
            transforms.CenterCrop(CONFIG["crop_size"]),
            transforms.ToTensor(),
            transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
        ]
    )

    image = Image.open(image_path).convert("RGB")
    image_tensor = transform(image).unsqueeze(0).to(device)

    # Encode image
    encoder.eval()
    decoder.eval()

    with torch.no_grad():
        features = encoder(image_tensor)

        # Generate caption using greedy decode
        start_token = tokenizer.word2idx["<start>"]
        end_token = tokenizer.word2idx["<end>"]

        generated_tokens = decoder.greedy_decode(
            features, start_token, end_token, max_length
        )[0]

        # Convert token indices to words
        caption_words = [
            tokenizer.idx2word[idx] for idx in generated_tokens[1:] if idx != end_token
        ]  # Skip <start>, stop at <end>
        caption_text = " ".join(caption_words)

        return (
            caption_text,
            None,
            image,
        )  # Return None for attention weights since not implemented


print("Inference function defined successfully!")
```

```python
# Data synthesis for CLIP demonstration
print("=== CLIP Data Synthesis ===")


# Create synthetic image-text pairs for CLIP training
def create_clip_dataset(num_samples=2000):
    """Create synthetic image-text pairs for CLIP training demonstration"""

    # Define categories and their descriptions
    categories = {
        "cat": ["a cat", "feline", "domestic cat", "tabby cat"],
        "dog": ["a dog", "canine", "puppy", "dog breed"],
        "car": ["a car", "automobile", "vehicle", "sedan"],
        "bird": ["a bird", "avian", "flying bird", "bird species"],
        "tree": ["a tree", "plant", "forest tree", "oak tree"],
    }

    data = []

    for i in range(num_samples):
        # Randomly select category
        category = np.random.choice(list(categories.keys()))

        # Create synthetic image path
        image_path = f"synthetic_{category}_{i:04d}.jpg"

        # Create corresponding text description
        base_desc = np.random.choice(categories[category])

        # Add variations
        variations = [
            f"photo of {base_desc}",
            f"image showing {base_desc}",
            f"picture of {base_desc}",
            f"{base_desc} in the scene",
        ]
        text = np.random.choice(variations)

        data.append({"image": image_path, "text": text, "category": category})

    return pd.DataFrame(data)


# Create CLIP dataset
clip_df = create_clip_dataset(2000)
print(f"Created CLIP dataset: {len(clip_df)} samples")
print(f"Categories: {clip_df['category'].value_counts().to_dict()}")

# Create synthetic images directory for CLIP
clip_images_dir = project_root / "q1_clip" / "synthetic_images"
clip_images_dir.mkdir(exist_ok=True)


def create_synthetic_image_clip(category, save_path):
    """Create a simple synthetic image based on category for CLIP"""

    # Color mapping for categories
    colors = {
        "cat": [255, 165, 0],  # Orange
        "dog": [0, 0, 255],  # Blue
        "car": [255, 0, 0],  # Red
        "bird": [0, 255, 0],  # Green
        "tree": [139, 69, 19],  # Brown
    }

    # Create colored square image
    img = np.full((224, 224, 3), colors.get(category, [128, 128, 128]), dtype=np.uint8)

    # Add some noise for texture
    noise = np.random.randint(-30, 30, (224, 224, 3))
    img = np.clip(img + noise, 0, 255).astype(np.uint8)

    pil_img = Image.fromarray(img)
    pil_img.save(save_path)


# Create synthetic images
print(f"Creating synthetic images in: {clip_images_dir}")
for i, row in tqdm(clip_df.iterrows(), total=len(clip_df), desc="Creating images"):
    img_path = clip_images_dir / row["image"]
    create_synthetic_image_clip(row["category"], img_path)

print("CLIP dataset created successfully!")
```

## 14. CLIP Training

Train the CLIP model using InfoNCE loss.


```python
# CLIP Training Implementation
print("=== CLIP Training ===")


class CLIPDataset(torch.utils.data.Dataset):
    def __init__(self, df, images_dir, vocab, transform=None):
        self.df = df
        self.images_dir = images_dir
        self.vocab = vocab
        self.transform = transform

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        img_path = self.images_dir / row["image"]

        # Load image
        image = Image.open(img_path).convert("RGB")
        if self.transform:
            image = self.transform(image)

        # Tokenize text
        text_tokens = []
        for word in row["text"].split():
            text_tokens.append(self.vocab.get(word, self.vocab["<unk>"]))

        # Pad/truncate to fixed length
        max_len = 10
        if len(text_tokens) < max_len:
            text_tokens += [self.vocab["<pad>"]] * (max_len - len(text_tokens))
        else:
            text_tokens = text_tokens[:max_len]

        return image, torch.tensor(text_tokens), row["category"]


# Create CLIP dataset and dataloader
clip_transform = transforms.Compose(
    [
        transforms.Resize(224),
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
    ]
)

clip_dataset = CLIPDataset(clip_df, clip_images_dir, clip_vocab, clip_transform)
clip_train_size = int(0.8 * len(clip_dataset))
clip_val_size = len(clip_dataset) - clip_train_size
clip_train_dataset, clip_val_dataset = torch.utils.data.random_split(
    clip_dataset,
    [clip_train_size, clip_val_size],
    generator=torch.Generator().manual_seed(SEED),
)

clip_train_loader = DataLoader(
    clip_train_dataset, batch_size=CLIP_CONFIG["batch_size"], shuffle=True
)
clip_val_loader = DataLoader(
    clip_val_dataset, batch_size=CLIP_CONFIG["batch_size"], shuffle=False
)

# Training loop
clip_optimizer = optim.Adam(
    [
        {"params": image_encoder.parameters()},
        {"params": text_encoder.parameters()},
        {"params": clip_loss_fn.parameters()},
    ],
    lr=CLIP_CONFIG["learning_rate"],
)
clip_losses = []

print("Starting CLIP training...")
for epoch in range(CLIP_CONFIG["num_epochs"]):
    image_encoder.train()
    text_encoder.train()
    clip_loss_fn.train()

    epoch_loss = 0

    for images, text_tokens, _ in tqdm(clip_train_loader, desc=f"CLIP Epoch {epoch+1}"):
        images = images.to(device)
        text_tokens = text_tokens.to(device)

        clip_optimizer.zero_grad()

        image_features = image_encoder(images)
        text_features = text_encoder(text_tokens)
        loss = clip_loss_fn(image_features, text_features)

        loss.backward()
        clip_optimizer.step()

        epoch_loss += loss.item()

    avg_loss = epoch_loss / len(clip_train_loader)
    clip_losses.append(avg_loss)

    print(f"Epoch {epoch+1}/{CLIP_CONFIG['num_epochs']}, Loss: {avg_loss:.4f}")

print("CLIP training completed!")
```

## 15. CLIP Evaluation

Evaluate CLIP on image-text retrieval tasks.


```python
# CLIP Evaluation
print("=== CLIP Evaluation ===")


def evaluate_clip_retrieval(image_encoder, text_encoder, dataloader, k=5):
    """Evaluate CLIP on retrieval tasks"""
    image_encoder.eval()
    text_encoder.eval()

    image_features = []
    text_features = []
    categories = []

    with torch.no_grad():
        for images, text_tokens, cats in dataloader:
            images = images.to(device)
            text_tokens = text_tokens.to(device)

            img_feat = image_encoder(images)
            txt_feat = text_encoder(text_tokens)
            image_features.append(img_feat.cpu())
            text_features.append(txt_feat.cpu())
            categories.extend(cats)

    image_features = torch.cat(image_features)
    text_features = torch.cat(text_features)

    # Image-to-text retrieval
    similarities = torch.matmul(image_features, text_features.t())

    # Top-k accuracy
    _, top_k_indices = similarities.topk(k, dim=1)

    correct = 0
    total = 0
    for i in range(len(similarities)):
        if i in top_k_indices[i]:
            correct += 1
        total += 1

    return correct / total


# Evaluate CLIP
clip_accuracy = evaluate_clip_retrieval(
    image_encoder, text_encoder, clip_val_loader, k=5
)
print(f"CLIP Top-5 Retrieval Accuracy: {clip_accuracy:.4f}")

# Visualize embeddings
print("=== CLIP Embedding Visualization ===")

# Get embeddings for visualization
image_encoder.eval()
text_encoder.eval()

sample_images, sample_texts, sample_categories = next(iter(clip_val_loader))
sample_images = sample_images[:100].to(device)
sample_texts = sample_texts[:100].to(device)

with torch.no_grad():
    img_embeds = image_encoder(sample_images)
    txt_embeds = text_encoder(sample_texts)

# Combine and reduce dimensionality
all_embeds = torch.cat([img_embeds, txt_embeds])
tsne = TSNE(n_components=2, random_state=SEED)
reduced_embeds = tsne.fit_transform(all_embeds.cpu().numpy())

# Plot
plt.figure(figsize=(10, 8))
categories = sample_categories[:100] * 2  # Duplicate for image and text
colors = ["red", "blue", "green", "orange", "purple"]
category_colors = [
    colors[list(set(sample_categories)).index(cat)] for cat in categories
]

plt.scatter(
    reduced_embeds[:100, 0],
    reduced_embeds[:100, 1],
    c=category_colors[:100],
    marker="o",
    label="Images",
    alpha=0.7,
)
plt.scatter(
    reduced_embeds[100:, 0],
    reduced_embeds[100:, 1],
    c=category_colors[100:],
    marker="s",
    label="Texts",
    alpha=0.7,
)

plt.title("CLIP Embeddings t-SNE Visualization")
plt.legend()
plt.savefig(
    output_dir / f"q1_clip_embeddings_{timestamp}.png", dpi=300, bbox_inches="tight"
)
plt.show()

print(
    f"CLIP embeddings visualization saved to: {output_dir / f'q1_clip_embeddings_{timestamp}.png'}"
)
```

```python
# Self-contained simple CLIP implementation (inline, no external imports)
class ImageEncoderSimple(nn.Module):
    def __init__(self, proj_dim=512):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(3, 64, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),
            nn.Conv2d(64, 128, 3, padding=1),
            nn.ReLU(),
            nn.AdaptiveAvgPool2d((1, 1)),
        )
        self.proj = nn.Linear(128, proj_dim)

    def forward(self, x):
        h = self.conv(x).view(x.size(0), -1)
        z = self.proj(h)
        z = z / (z.norm(dim=1, keepdim=True) + 1e-8)
        return z


class TextEncoderSimple(nn.Module):
    def __init__(self, vocab_size, proj_dim=512, emb_dim=128):
        super().__init__()
        self.emb = nn.Embedding(vocab_size, emb_dim)
        self.pool = nn.AdaptiveAvgPool1d(1)
        self.proj = nn.Linear(emb_dim, proj_dim)

    def forward(self, tokens):
        e = self.emb(tokens).permute(0, 2, 1)  # B x emb_dim x L
        h = self.pool(e).squeeze(2)
        z = self.proj(h)
        z = z / (z.norm(dim=1, keepdim=True) + 1e-8)
        return z


class InfoNCE(nn.Module):
    def __init__(self, temperature=0.07):
        super().__init__()
        self.logtau = nn.Parameter(torch.log(torch.tensor(temperature)))

    def forward(self, img_z, txt_z):
        tau = torch.exp(self.logtau)
        logits = img_z @ txt_z.t() / tau
        labels = torch.arange(img_z.size(0)).to(img_z.device)
        loss_i2t = nn.CrossEntropyLoss()(logits, labels)
        loss_t2i = nn.CrossEntropyLoss()(logits.t(), labels)
        return 0.5 * (loss_i2t + loss_t2i)


# Small demo setup
CLIP_CONFIG = {"embed_dim": 256, "batch_size": 64, "lr": 1e-4, "num_epochs": 3}
image_encoder_clip = ImageEncoderSimple(proj_dim=CLIP_CONFIG["embed_dim"]).to(device)
text_encoder_clip = TextEncoderSimple(
    vocab_size=1000, proj_dim=CLIP_CONFIG["embed_dim"]
).to(device)
clip_loss = InfoNCE(temperature=0.07).to(device)
print("Inline CLIP models initialized")
```
